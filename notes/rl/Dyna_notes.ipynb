# Dyna（Sutton, 1990）笔记（速查版）

> 关键词：**Direct RL（真实交互更新）**、**Model Learning（学环境模型）**、**Planning（想象/模拟更新）**、**Dyna-Q**、**Model Bias（模型偏差）**  
> 一句话：**用真实数据学 + 同时学模型 + 用模型生成虚拟经验再学**。

---

## 1. Dyna 解决了什么问题？（为什么出现）

### 场景
- **真实环境交互昂贵/慢/有风险**（机器人、工业控制、线上系统）
- 纯 **Model-free**（如 Q-learning/PPO）往往 **样本效率低**
- 纯 **Model-based/规划**需要较准模型，否则易偏

### 目标
- 在 **Model-free 的简单稳定** 和 **Model-based 的高样本效率** 之间折中  
- “每条真实经验都要榨干”：真实更新 + 多次虚拟更新

---

## 2. Dyna 的三大组件（框架）

1) **Direct RL（直接学习）**  
   用真实交互 \((s,a,r,s')\) 更新价值/策略（例如 Q-learning 一步）

2) **Model Learning（学模型）**  
   从真实数据学习（或记忆）环境模型：  
   \[
   \hat P(s'|s,a),\quad \hat R(s,a)
   \]
   （表格/回归网络/世界模型都行）

3) **Planning（规划/想象回放）**  
   从已见过的 \((s,a)\) 采样，用模型生成 \((\hat r, \hat s')\) 做额外更新  
   → 相当于“脑内练习”N 次

---

## 3. Dyna-Q 典型流程（直觉版）

每个真实环境 step 后：
1. 真环境走一步：得到 \((s,a,r,s')\)
2. 用这条真实数据做一次 **Q-learning 更新**
3. 更新模型（存/拟合 \(\hat P,\hat R\)）
4. **Planning N 次**：从过去见过的 \((\tilde s,\tilde a)\) 抽样，用模型生成 \((\hat r, \hat s')\)，再做 N 次 Q-learning 更新

> **超参 N**：每次真实交互后做多少次模型生成的“虚拟更新”。

---

## 4. N（planning 次数）的 trade-off（必考点）

- **N 越大：**
  - ✅ 好处：样本效率更高、学得更快（真实数据被重复利用更多）
  - ❗ 风险：模型不准时 **Model Bias 放大**（越 planning 越被带偏）
  - ❗ 代价：计算开销更大

---

## 5. Dyna vs Experience Replay（经验回放）

- **相同点**：都做“数据复用”，提高稳定性/样本效率
- **不同点**：
  - Replay：重放的是 **真实经验**
  - Dyna planning：使用 **模型生成的虚拟经验**（更省样本，但引入模型偏差）

---

## 6. 什么时候 Dyna 特别划算？（适用条件）

1. **交互成本高**（真实试错贵）  
2. **环境可建模、模型能学得较准**（动力学可预测、噪声不离谱）  
3. **N 选得合理**：放大样本利用率但不过度放大模型误差

---

## 7. 典型坑（模型难学准时会怎样）

- **模型误差累积** → 虚拟数据系统性错 → 更新方向被带偏  
- **越 planning 越糟**（N 大时偏差放大，甚至发散）  
- **策略被模型“骗”**：在模型里很赚，真实环境不行（classic failure mode）

---

## 8. 记忆卡片（只背这 4 行）

- Dyna = **真实更新 + 学模型 + 模型模拟更新（planning）**
- Dyna-Q：每个真实 step 后做 **N 次**模型生成的 Q 更新
- N ↑：样本效率 ↑；模型偏差风险 ↑；计算 ↑
- Replay 用真实经验；Dyna planning 用模型生成经验
